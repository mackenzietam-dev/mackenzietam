```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("noisy_curvefit_data.csv")
df
```

```python
print("Shape:", df.shape)
print("First five rows:")
print(df.head())
print("Last five rows:")
print(df.tail())
print("Columns:", df.columns)
```

```python
x = df["x"]
y_noise_1 = df["y_noise_1"]

plt.scatter(x, y_noise_1, color="gray", label="Noise level 1")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Noise Level 1")
plt.legend()
plt.show()
```

```python
def model(x, A, k):
    return A * np.exp(-k * x)

p0 = [1, 1]

popt, pcov = curve_fit(model, x, y_noise_1, p0=p0)

A_fit, k_fit = popt

print(f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} * x)")

residuals = y_noise_1 - model(x, A_fit, k_fit)
chi2 = np.sum(residuals**2)
dof = len(x) - len(popt)
chi2_reduced = chi2 / dof

print("Reduced chi-squared:", round(chi2_reduced, 2))
```

```python
plt.scatter(x, y_noise_1, color="gray", label="Noisy data")
plt.plot(x, model(x, A_fit, k_fit), color="red", label="Fitted model")
plt.plot(x, model(x, 5, 0.8), color="black", linestyle="--", label="True model")

plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Noise Level 1, Reduced Chi-Squared = {chi2_reduced:.2f}")
plt.legend()
plt.show()
```

```python
def fit_model(x, y):
    popt, pcov = curve_fit(model, x, y, p0=[1, 1])

    A_fit, k_fit = popt

    residuals = y - model(x, A_fit, k_fit)
    chi2 = np.sum(residuals**2)
    dof = len(x) - len(popt)
    chi2_reduced = chi2 / dof

    return popt, chi2_reduced
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for i in range(2, 6):
    y = df[f"y_noise_{i}"]

    popt, chi2_reduced = fit_model(x, y)
    A_fit, k_fit = popt

    ax = axes.flat[i - 2]

    ax.scatter(x, y, color="gray", label="Noisy data")
    ax.plot(x, model(x, A_fit, k_fit), color="red", label="Fitted model")
    ax.plot(x, model(x, 5, 0.8), color="black", linestyle="--", label="True model")

    ax.set_xlabel("x")
    ax.set_ylabel("y")
    ax.set_title(f"Noise Level {i}, Reduced Chi-Squared = {chi2_reduced:.2f}")
    ax.legend()

plt.tight_layout()
plt.show()
```

1.

As the noise increases, the red fitted line generally becomes less accurate and moves farther away from the black dotted line. This happens because larger random variations make it harder to find the original relationship in the data. 

2.

Decreased chi-squared generally increases as the noise level increases. This happens because the data points are usually farther away from the fitted curve which makes the sum of squared residuals larger.


```python
results = []

for i in range(1, 6):
    y = df[f"y_noise_{i}"]

    popt, chi2_reduced = fit_model(x, y)
    A_fit, k_fit = popt

    results.append([i, A_fit, k_fit, chi2_reduced])

results_df = pd.DataFrame(results, columns=[
    "Noise Level",
    "A fit",
    "k fit",
    "Reduced Chi-Squared"
])

print(results_df)

results_df.to_csv("noisy_parameters.csv", index=False)
```
