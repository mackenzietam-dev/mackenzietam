## 3.1 - 1d numpy arrays

```python
import numpy as np

numbers = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(numbers)

print("multiplied by 4:", arr * 4)
print("mean:", np.mean(arr))
print("median:", np.median(arr))
print("standard deviation:", np.std(arr))

smallest_index = np.argmin(arr)
largest_index = np.argmax(arr)
print("smallest index:", smallest_index)
print("largest index:", largest_index)
print("smallest value using index:", arr[smallest_index])
print("largest value using index:", arr[largest_index])
print("smallest value using np.min:", np.min(arr))
print("largest value using np.max:", np.max(arr))
print("do they match?", arr[smallest_index] == np.min(arr) and arr[largest_index] == np.max(arr))
print("values greater than 20:", arr[arr > 20])
```

## 3.2 - 2d numpy arrays

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

print("shape:", matrix.shape)
print("rows:", matrix.shape[0])
print("columns:", matrix.shape[1])
print("third column:", matrix[:, 2])
print("second row:", matrix[1, :])
print("multiplied by 2:\n", matrix * 2)
print("mean of each column:", np.mean(matrix, axis=0))
print("overall mean:", np.mean(matrix))
print("max of each row:", np.max(matrix, axis=1))
print("min of each column:", np.min(matrix, axis=0))

flat = matrix.flatten()
print("flattened array:", flat)
print("new shape:", flat.shape)
print("what happened: the rows were joined into one 1d array")

smallest_flat_index = np.argmin(matrix)
largest_flat_index = np.argmax(matrix)
print("smallest flattened index:", smallest_flat_index)
print("largest flattened index:", largest_flat_index)
print("smallest row-column coordinates:", np.unravel_index(smallest_flat_index, matrix.shape))
print("largest row-column coordinates:", np.unravel_index(largest_flat_index, matrix.shape))
print("values greater than 15:", matrix[matrix > 15])
print("each row sorted:\n", np.sort(matrix, axis=1))
```

## 4 - prime star systems

```python
stars_magnitudes = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True

def prime_star_systems(values):
    return values[np.array([any(is_prime(number) for number in row) for row in values])]

print(prime_star_systems(stars_magnitudes))
```

## 5 - checkers

### 5.1 - an 8x8 array of zeros

```python
def empty_checkerboard():
    return np.zeros((8, 8), dtype=int)

print(empty_checkerboard())
```

### 5.2 - odd-numbered rows starting with 1

```python
def odd_row_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 0::2] = 1
    return board

print(odd_row_checkerboard())
```

### 5.3 - completed checkerboard

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 0::2] = 1
    board[1::2, 1::2] = 1
    return board

print(checkerboard())
```

### 5.4 - reversed checkerboard

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[0::2, 1::2] = 1
    board[1::2, 0::2] = 1
    return board

print(reverse_checkerboard())
```

## 6 - the expanding universe

```python
def expansion(words, spaces):
    separator = " " * spaces
    return np.array([separator.join(word) for word in words])

universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```

## 7 - second-brightest star

```python
def secondDimmest(stars):
    return np.sort(stars, axis=1)[:, -2]

stars = np.array([
    [1123, 1456, 1789, 1324, 1876],
    [1567, 1987, 1678, 1405, 1589],
    [1345, 1654, 1523, 1109, 1923],
    [1298, 1890, 1367, 1784, 1432],
    [1823, 1756, 1489, 1672, 1550]
])
print(secondDimmest(stars))
```
